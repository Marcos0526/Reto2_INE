## Objetivo

Comprender la estructura, calidad y comportamiento de los datos de participación
electoral antes de construir el modelo de riesgo.

## Contexto

Conteos Censales de Participación 2024 del INE (Datos Abiertos).

## Fases asociadas

- Fase 1: Comprensión de la información
- Fase 2: Preparación de los datos

```python
import glob
import os

import pandas as pd

DATA_DIR = '../DatosAbiertos_DECEyEC_ConteosCensales2024'
ANIOS = [2024, 2021, 2018, 2015, 2012, 2009]

files = []
for anio in ANIOS:
    files += sorted(glob.glob(os.path.join(DATA_DIR, f'datosabiertos_deceyec_conteoscensales{anio}_*.csv')))

files = sorted(set(files))
print(len(files))
```

## 1. Metadatos

```python
contenido = open(os.path.join(DATA_DIR, '01_descripcioncampos_conteoscensales2024.txt'), 'r', encoding='utf-8').read()
print(contenido)
```

## 2. Carga de los datos

```python
frames = []
for f in files:
    frames.append(pd.read_csv(f))

df = pd.concat(frames, ignore_index=True)
print(df.shape)
df.head()
```

```python
df.info()
df.describe(include='all')
```

## 4. Indicadores derivados

```python
df['tasa_participacion'] = df['SV'] / (df['SV'] + df['NV'])
df['tasa_abstencion'] = df['NV'] / (df['SV'] + df['NV'])
tasa_por_entidad = df.groupby(['AELEC', 'EDONOM'])['tasa_abstencion'].mean().sort_values(ascending=False)
tasa_por_entidad
```
